```python
class cylinder():

    pi = 3.14
    def __init__(self,height = 1,radius = 1):
        self.height = height
        self.radius = radius

    def volume(self):
        return(self.pi*self.radius**2*self.height)

    def area(self):
        return(2*self.pi*self.radius*self.height+ 2*self.pi*self.radius**2)
```

```python
oval = cylinder(6,8)
```

```python
class Line():
    
    def __init__(self,coor1,coor2):
        self.coor1 = coor1
        self.coor2 = coor2
    
    def distance(self):
        x1 , y1 = self.coor1
        y2 , x2 = self.coor2

        return(((x1 - x2)**2 + (y1 - y2)**2)**0.5)
    
    def slope(self):

       x1 , y1 = self.coor1
       y2 , x2 = self.coor2

       return((y2 - y1)/(x2 - x1))
        
```

```python
a = (7,8)
b = (9,10)
```

```python
myline = Line(a,b)
```

```python
myline.distance()
```

```python
myline.slope()
```

```python
class Account():
    def __init__(self,owner,balance):
        self.owner = owner
        self.balance = balance

    def deposit(self,dep_amount):
        self.balance + dep_amount
        print(f'added {dep_amount} to the balance')
        

    def withdrawl(self,wd_amount):

        if self.balance >= wd_amount:
            self.balance = self.balance - wd_amount
            print('withdrawl accepted')
        else:
            print('sorry not have enough balance')

    def __str__(self):
        return(f'owner: {self.owner} \nbalance: {self.balance}')
```

```python
class deck():
    def __init__(self):

        self.all_cards = []

        for suit in suits:
            for rank in ranks:
                #create the card object
                created_card = card(suit,rank)
                self.all_cards.append(created_card)

    def shuffle(self):
        random.shuffle(self.all_cards)

    def deal_one(self):
        return self.all_cards.pop()
```

```python
class player:

    def __init__(self,name):
        self.name = name
        self.all_cards = []

    def remove_one(self):
        return self.all_cards.pop(0)

    def add_cards(self,new_cards):
        #list of multiple cards
        if type(new_cards) == type([]):
            self.all_cards.extend(new_cards)
        else:
            #list of single cards
            self.all_cards.append(new_cards)

    def __str__(self):
        return f'player {self.name} has {len(self.all_cards)} cards.'
```
